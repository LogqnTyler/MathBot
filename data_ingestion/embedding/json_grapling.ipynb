```python
import json
from dataclasses import dataclass
```

```python
PATH = "/Users/loganfoster/projects/MathBot/JSON/lesson14.json"

data = json.load(open(PATH, "r"))
```

```python
@dataclass
class Chunk:
    kind: str
    mat_id: int ## this is the reference to the lesson file
    idx: int
    name: str = ""
    learning_objective: int | None = None
    content_plain: str = ""
    content_latex: str = ""
    prob_context_plain: str = ""
    prob_context_latex: str = ""
    sub_prob_part: str = ""
    Q_plain: str = ""
    Q_latex: str = ""
    A_plain: str = ""
    A_latex: str = ""
    keywords: list[str] = None
```

```python
def parse_contents(contents: dict, mat_id = 69) -> list[dict]:
    keys = contents.keys()
    chunks = []
    for key in keys:
        if key == "definitions":
            for idx, definition in enumerate(contents[key]):
                defined_term = definition["term"]
                chunk = Chunk(
                    kind="definition",
                    name = defined_term,
                    mat_id=mat_id,
                    idx=idx,
                    content_plain=f"Def: {defined_term} \n" + definition["definition_plain"],
                    content_latex=f"Def: {defined_term} \n" + definition["definition_latex"],
                    keywords=[defined_term]
                )
                chunks.append(chunk)
        elif key == "other_material":
            for idx, material in enumerate(contents[key]):
                kind = material["type"]
                chunk = Chunk(
                    kind=kind,
                    name = kind + f"_{idx}",
                    mat_id=mat_id,
                    idx=idx,
                    content_plain=material["content_plain"],
                    content_latex=material["content_latex"],
                )
                chunks.append(chunk)
        elif key == "problems":
            for idx, problem in enumerate(contents[key]):
                context_plain = problem["context_plain"]
                context_latex = problem["context_latex"]
                name = problem["name"] + f"_{idx}_{sub_prob["part"]}" ## ensures unique name for each problem
                keywords = problem["keywords"]
                for sub_prob in problem["subproblems"]:
                    chunk = Chunk(
                        kind="problem",
                        name = name,
                        mat_id=mat_id,
                        idx=idx,
                        prob_context_plain=context_plain,
                        prob_context_latex=context_latex,
                        sub_prob_part=sub_prob["part"],
                        Q_plain=sub_prob["plain_text"]["question"],
                        Q_latex=sub_prob["latex"]["question"],
                        A_plain=sub_prob["plain_text"]["answer"],
                        A_latex=sub_prob["latex"]["answer"],
                        keywords=keywords
                    )
                    chunks.append(chunk)
    return chunks
```

```python
chunks = parse_contents(data["contents"])
```
